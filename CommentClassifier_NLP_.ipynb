```python
path = untar_data(URLs.IMDB)

path.ls()
```

```python
dls = TextDataLoaders.from_folder(path, train='train', valid='test')

learn = text_classifier_learner(dls, AWD_LSTM, drop_mult=0.5, metrics = accuracy)

learn.fine_tune(4, 0.01)
```
